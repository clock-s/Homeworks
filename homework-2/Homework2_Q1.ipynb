## **Item 2**

```r
n <- 50
p <- 0.7

pessoas <- seq(0, n, 1)
BIN_PMF <- dbinom(pessoas, n, p)


plot(pessoas, BIN_PMF)

BIN_CDF <- c(cumsum(BIN_PMF))

plot(pessoas, BIN_CDF)
```
